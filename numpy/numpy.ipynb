#numpy array와 matrix의 차이가 뭐야
Numpy ndarray와 matrix는 둘 다 행렬 형태의 데이터를 다루지만, ndarray가 Numpy의 기본 데이터 구조이며, matrix는 특정한 행렬 연산을 위해 만들어진 클래스입니다.

1. 차원 (Dimension)
ndarray: 1차원, 2차원, 3차원 등 다차원(N-dimensional) 배열을 표현할 수 있습니다.
matrix: 오직 2차원 행렬만 표현할 수 있습니다.

1. 연산자 (Operator)
이것이 가장 큰 차이점입니다.

* 연산자:
ndarray: 요소별 곱셈(element-wise multiplication)을 수행합니다. 즉, 두 배열의 같은 위치에 있는 요소끼리 곱합니다. 행렬 곱셈을 하려면 @ 연산자 또는 np.dot() 함수를 사용해야 합니다.
matrix: 기본적으로 **행렬 곱셈(matrix multiplication)**을 수행합니다.

3. 전치 (Transpose)
ndarray: .T 속성을 사용해 전치할 수 있습니다. 예를 들어 arr.T.
matrix: .T 외에 .H(Hermitian transpose, 켤레 복소 전치)도 지원합니다.

4. 권장 사항
ndarray: 대부분의 상황에서 ndarray 사용을 권장합니다. matrix는 특정 용도를 위해 존재하지만, ndarray의 기능이 더 풍부하고 현대적인 Numpy 생태계와 더 잘 맞습니다. Numpy 문서에서도 matrix보다는 ndarray를 사용하라고 권장하고 있습니다.
matrix: ndarray와 matrix의 연산 방식 차이 때문에 혼란이 생길 수 있으므로, 행렬 연산이 필요한 경우 ndarray에 @ 연산자를 사용하는 것이 더 명확하고 일관적입니다.

```python
import numpy as np

arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[5, 6], [7, 8]])

mat1 = np.matrix([[1, 2], [3, 4]])
mat2 = np.matrix([[5, 6], [7, 8]])

# ndarray는 요소별 곱셈
print(f"ndarray * : \n{arr1 * arr2}\n")

# matrix는 행렬 곱셈
print(f"matrix * : \n{mat1 * mat2}\n")

# ndarray로 행렬 곱셈을 하려면 @ 또는 dot() 사용
print(f"ndarray @ : \n{arr1 @ arr2}\n")
```
